#### 3. Call the OpenAI API with different `temperature` values

`temperature` values: `0`, `0.7` and `1.5`. Generate three responses at each temperature.

Use `gpt-5.4-mini` for this task and set `reasoning_effort="none"`.

Compare the outputs.

Tip: With Chat Completions API you can use parameter `n=3` to generate three different responses to your prompt with one call.

````python
prompt3 = "Write a two sentence story about a robot learning to cook."

response_t0 = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": prompt3}],
    temperature=0,
    n=3,
    reasoning_effort="none"
)

for i, choice in enumerate(response_t0.choices, start=1):
    print(f"--- Response {i} ---")
    print(choice.message.content)
````

#### 18. Generate a Python code with API call.

Write a prompt that generates a Python function to calculate the factorial of a number. Once you have written the prompt, use it to call the API and generate the code. Test the generated code to ensure it calculates the factorial correctly for various inputs.

````python
prompt18 = (
    "Write a clean Python function named `calculate_factorial(n)` that calculates the factorial of a non-negative integer. "
    "Include input validation for negative numbers. Return only the Python code without markdown formatting."
)

res_code = client.responses.create(
    model="gpt-5.4-mini",
    input=prompt18
)

print(res_code.output_text)
````

````python
def calculate_factorial(n):
    if not isinstance(n, int):
        raise TypeError("n must be an integer")
    if n < 0:
        raise ValueError("n must be a non-negative integer")
    
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print("0! =", calculate_factorial(0))   # Must be 1
print("1! =", calculate_factorial(1))   # Must be 1
print("5! =", calculate_factorial(5))   # Must be 120
print("7! =", calculate_factorial(7))   # Must be 5040
````

Extract the information from the text above using two different methods:

2. Instead of putting rules in the prompt, provide a specific JSON schema that defines the field names and their required data types `destination` (the first city), `start_date` (formatted as YYYY-MM-DD), and `stay_duration` (the number of nights as an integer).

Which method yields a more reliable result for a computer program to process, and why?

````python
input_text = "Our trip to Tokyo starts on June 12th. We're staying for 5 nights at the Park Hyatt before heading to Kyoto."

res_json_prompt = client.responses.create(
    model="gpt-5.4-mini",
    input=(
        "Extract the following fields from the text and return them as a valid JSON object: "
        "`destination`, `start_date`, and `stay_duration`.\n\n"
        f"Text: {input_text}"
    )
)

print(res_json_prompt.output_text)
````

````python
import json

trip_schema = {
    "type": "object",
    "properties": {
        "destination": {
            "type": "string",
            "description": "The first city visited on the trip."
        },
        "start_date": {
            "type": "string",
            "description": "The start date formatted strictly as YYYY-MM-DD."
        },
        "stay_duration": {
            "type": "integer",
            "description": "The number of nights staying as an integer."
        }
    },
    "required": ["destination", "start_date", "stay_duration"],
    "additionalProperties": False
}

res_json_schema = client.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[
        {"role": "system", "content": "Extract trip details from the user's text. If year is not mentioned, assume 2026."},
        {"role": "user", "content": input_text}
    ],
    response_format={
        "type": "json_schema",
        "json_schema": {
            "name": "trip_extraction",
            "strict": True,
            "schema": trip_schema
        }
    }
)

print(res_json_schema.choices[0].message.content)
````

#### Conclusion
More reliable results for a computer to process is the second JSON schema method because it will always have the same results without additional context.
