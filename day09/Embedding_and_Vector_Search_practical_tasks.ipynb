```python
# place your code here
import numpy as np

def numpy_cos_sim (vector, vector_batch):
    return np.round(np.dot(vector_batch, vector) / (np.linalg.norm(vector) * np.linalg.norm(vector_batch, axis=1)), 4)

numpy_cos_sim(embedding, embedding_list)
```
